```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(4,3))
plt.plot(x,y,marker='o',linestyle='--')
plt.title("Test Chart")
plt.xlabel('X1')
plt.ylabel('Y1')
plt.show()
```

```python
x= [1,2,3,4,5]
y1 = [10,20,25,35,45]
y2 = [20,30,40,55,42]


plt.figure(figsize=(4,3))
plt.plot(x,y1,marker='o',linestyle='--',label='Sale 2024')
plt.plot(x,y2,label='Sale 2025')
plt.title("YoY Sales")
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
#Bar chart
x=['Apple','Banana','Ornage','Cherry']
y=[10,20,30,5]

plt.bar(x,y)
plt.show()
```

```python
#Pie Chart

x= ['India','USA','Canada','UK']
y=[20,30,10,40]

plt.pie(y,labels=x,autopct='%1.1f%%')
plt.show()
```
