Linear Regression on Diabetes

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
```

```python
data = load_diabetes()
print (data['DESCR'])
```

```python
data.keys()
```

```python
feature = pd.DataFrame(data.data,columns=data.feature_names)
feature
```

```python
target = pd.Series(data.target, name="target")
target
```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
lr = LinearRegression()
```

```python
# Single Regression prod - Only one feature
columns = ['age','bmi','bp']
for i in columns:
  plt.scatter(feature[i],target)
  plt.xlabel(i)
  plt.ylabel('Target')
  plt.show()
```

```python
X = feature['bmi'].values.reshape(442,1)

print(X.shape)
```

```python
y = target
```

```python
lr.fit(X,y)
lr.coef_ #Slope - How much Y changes when X is increase by 1 unit

```

```python
lr.intercept_ # Value of Y when x=0
```

```python
pred = lr.predict(X)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#evaluate the model
mse = mean_squared_error(y,pred)
mse
```

```python
r2 = lr.score(X,y)
r2
```
