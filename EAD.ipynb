```python
import pandas as pd
import numpy as np

data = {'Name': ['Alice', 'Bob','Neha'], 'weight': [55, 70,np.nan],'age':[25,30,28]}
df = pd.DataFrame(data)
print(df)
```

```python
df['weight'] = df['weight'].fillna(df['weight'].mean())
df
```

```python
# Scaling / Feature Scaling
# Bringing all the numeric features to a similar scale
# -> Standardization - Convert numbers so that they are measured related to the average, the mean becomes 0 and STD becomes 1(Z- Score)
# Z = (x-mean)/ Standard Deviation
# When to use - Normally distributted data (60,65,70,75,67,84)
# Algo : Linear Regression/Logistic Regression/KNN/K-Mean/PCA/SupportsDivMod
# -> Normalization
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
std = StandardScaler()
df[['weight','age']] = std.fit_transform(df[['weight','age']])
df
```

## Normalization
# -> Fixed Range = 0 to 1
X_normalized = (x-min)/(max-min)
Not sensitive to outliers
When to used Normalization
-> Data is not normally distributed
- Alg0 - KNN/K -Mena/ Neural Networks

```python
# Simpleimputor
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='mean')
df[['weight','age']] = imputer.fit_transform(df[['weight','age']])
df
```

```python
scaler = MinMaxScaler()
df[['weight','age']] = scaler.fit_transform(df[['weight','age']])
df.head()
```

# Categorical Variables
# T-Shirt

1.   Color -> red,yellow,green,bluew,white
2.   Size -> XS,S,M,L,XL
3. Price -> 10,20,12,30

#### Two types of categorical variables
- Ordinal Categorical variables --> which can be ordered XS<S<M<L<XL
- Nominal Categorical variables --> Data which can't be ordered

```python
import pandas as pd

# Data for the categorical example
colors = ['red', 'yellow', 'green', 'blue', 'white']
sizes = ['XS', 'S', 'S', 'L', 'L']
prices = [10, 20, 12, 30, 15] # Added an extra price to match length
class_data= ['class1','class2','class3','class1','class3']
# Create the DataFrame
clothing_df = pd.DataFrame({
    'Color': colors,
    'Size': sizes,
    'Price': prices,
    'Class': class_data
})

print(clothing_df)
```

```python
size_mapping = {'XS':0,'S':1,'M':2,'L':3,'XL':4}
clothing_df['Size'] = clothing_df['Size'].map(size_mapping)
clothing_df
```

```python
# Label Encoding (Ordinal Data)
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
clothing_df['Size'] = le.fit_transform(clothing_df['Size'])
clothing_df
```
